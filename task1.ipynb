```python
import pandas as pd

df = pd.read_csv(
    "data/API_SP.POP.TOTL_DS2_en_csv_v2_33112/API_SP.POP.TOTL_DS2_en_csv_v2_33112.csv",
    skiprows=4
)

df.head()
```

```python
df.info
```

```python
df.info()
```

```python
# Remove the completely empty column
df = df.drop(columns=["Unnamed: 70"])
```

```python
df.shape
```

```python
df[["Country Name","2025"]].head(10)
```

```python
plt.figure(figsize=(10, 6))

plt.hist(df["2025"].dropna(), bins=30)

plt.title("Distribution of Population Across Countries/Regions in 2025")
plt.xlabel("Population")
plt.ylabel("Number of Countries/Regions")

plt.show()
```

```python
# Find the 10 most populated countries/regions in 2025
top10 = df[["Country Name", "2025"]].dropna().sort_values(
    by="2025", ascending=False
).head(10)

# Create a bar chart
plt.figure(figsize=(12, 6))

plt.bar(top10["Country Name"], top10["2025"])

plt.title("Top 10 Most Populated Countries/Regions in 2025")
plt.xlabel("Country/Region")
plt.ylabel("Population")

plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

```python
country_metadata = pd.read_csv(
    "data/API_SP.POP.TOTL_DS2_en_csv_v2_33112/Metadata_Country_API_SP.POP.TOTL_DS2_en_csv_v2_33112.csv",
    skiprows=0
)

country_metadata.head()
```

```python
country_codes = country_metadata[
    country_metadata["Region"].notna()
]["Country Code"]

# Keep only actual countries in our population dataset
countries_df = df[df["Country Code"].isin(country_codes)].copy()

# Check the result
countries_df.shape
```

```python
countries_df[["Country Name", "2025"]].head()
```

```python
top10_countries = (
    countries_df[["Country Name", "2025"]]
    .dropna()
    .sort_values(by="2025", ascending=False)
    .head(10)
)

# Display the data
top10_countries
```

```python
# Population distribution for 2025

plt.figure(figsize=(10, 6))

plt.hist(df["2025"].dropna(), bins=20)

plt.title("Population Distribution of Countries/Regions in 2025")
plt.xlabel("Population")
plt.ylabel("Number of Countries/Regions")

plt.show()
```
